## Byte Pair Encoding : a data compression and text tokenization algorithm used by most modern Large Language Models like GPT and RoBERTa

BPE algorithm operates by iteratively replacing the most common contiguous sequences of characters in a target text with unused 'placeholder' bytes. The iteration ends when no sequences can be found, leaving the target text effectively compressed. Decompression can be performed by reversing this process, querying known placeholder terms against their corresponding denoted sequence, using a lookup table. In the original paper, this lookup table is encoded and stored alongside the compressed text.

Example

Suppose the data to be encoded is:

<strong>aaabdaaabac</strong>

The byte pair "aa" occurs most often, so it will be replaced by a byte that is not used in the data, such as "Z".

Now there is the following data and replacement table:

<strong>ZabdZabac<br>
Z=aa</strong>

Then the process is repeated with byte pair "ab", replacing it with "Y":

<strong>ZYdZYac<br>
Y=ab<br>
Z=aa<br></strong>

The only literal byte pair left occurs only once, and the encoding might stop here. Alternatively, the process could continue with recursive byte-pair encoding, replacing "ZY" with "X":

<strong>
XdXac<br>
X=ZY<br>
Y=ab<br>
Z=aa<br>
</strong>

This data cannot be compressed further by byte-pair encoding because there are no pairs of bytes that occur more than once.

To decompress the data, simply perform the replacements in the reverse order.

```python
#Step 1: Created a text string for which we want to create the BPE
text = "Ｕｎｉｃｏｄｅ! 🅤🅝🅘🅒🅞🅓🅔‽ 🇺‌🇳‌🇮‌🇨‌🇴‌🇩‌🇪! 😄 The very name strikes fear and awe into the hearts of programmers worldwide. We all know we ought to “support Unicode” in our software (whatever that means—like using wchar_t for all the strings, right?). But Unicode can be abstruse, and diving into the thousand-page Unicode Standard plus its dozens of supplementary annexes, reports, and notes can be more than a little intimidating. I don’t blame programmers for still finding the whole thing mysterious, even 30 years after Unicode’s inception."

#Step 2: Convert that to the unicode encoding with utf-8
unicode_tokens = list(map(int,text.encode('utf-8')))

print("-----")
print(text)
print("text length : ",len(text))
print("-----")
print(unicode_tokens)
print("unicode lenth:",len(unicode_tokens))
```

```python
#Step 3: Count the maximum occuring pair from the list of unicodes
def get_stats(tokens):
  count = {}
  for pairs in zip(tokens,tokens[1:]):
    count[pairs] = count.get(pairs,0) + 1
  return count

pair_counter = get_stats(unicode_tokens)
#sort the pair counter based on the values - descending
pair_counter = dict(sorted(pair_counter.items(), key=lambda item: item[1],reverse=True))
print(pair_counter)
```

```python
#Implementing the next stage of iteration
def get_stats(tokens):
  count = {}
  for pairs in zip(tokens,tokens[1:]):
    count[pairs] = count.get(pairs,0) + 1
  return count

#Step 4: Merge the maximum occuring pair with a new tokenId
def merge_pairs(pair,unicode_tokens,new_tokenId):
  new_tokenIDs = []
  i = 0
  while i < len(unicode_tokens):
    #print(unicode_tokens[i])
    if i<(len(unicode_tokens)-1) and unicode_tokens[i] == pair[0] and unicode_tokens[i+1] == pair[1]:
      #print(pair[0],pair[1])
      new_tokenIDs.append(new_tokenId)
      i += 2
    else:
      new_tokenIDs.append(unicode_tokens[i])
      i += 1
  return new_tokenIDs

vocab_size = 276
number_of_merges = vocab_size - 256
intitial_tokens = list(map(int,text.encode('utf-8')))
new_tokens= list(map(int,text.encode('utf-8')))

for i in range(number_of_merges):
  pair_counter = get_stats(new_tokens)
  max_occuring_pair = max(pair_counter, key=pair_counter.get)
  print(max_occuring_pair," merged into token ",256+i)
  new_tokens = merge_pairs(max_occuring_pair,new_tokens,256+i)
```
